### Data cleaning 

```python
sns.boxplot(x='Pclass', y='Age', data=titanic)
plt.show()
```

```python
# Python programme
def input_age(cols):
    Age=cols[0]
    Pclass=cols[1]

    if pd.isnull(Age):
        if Pclass==1:
            return 37
        elif Pclass==2:
            return 29
        else:
            return 24
    else:
        return Age
```

```python
titanic['Age']=titanic[['Age','Pclass']].apply(input_age, axis=1)
```

```python
sns.heatmap(titanic.isnull(), yticklabels=False, cbar=False, cmap='viridis')
plt.show()
```

```python
# in cabin column there are a lot of null value which can be clean using feature engineering jo abhi drop kr rre hia 
titanic.drop('Cabin', axis=1, inplace=True)  # delete the Cabin column 
```

```python
titanic  # check that Cabin column has been droped 
```

**Now Data has been cleaned**

```python
sns.heatmap(titanic.isnull(), yticklabels=False, cbar=False, cmap='viridis')
plt.show()
# This indicate that there is no null value present in the data
```

### Converting Categorial features 

```python
titanic.info()
```

```python
pd.get_dummies(titanic['Embarked'], drop_first=True).head()
```

```python
embark=pd.get_dummies(titanic['Embarked'], drop_first=True)
sex=pd.get_dummies(titanic['Sex'], drop_first=True)
```

```python
titanic.drop(['Name', 'Embarked', 'Sex', 'Ticket'], axis=1, inplace=True)
```

```python
titanic
```

```python
titanic=pd.concat([titanic, sex, embark], axis=1)
```

```python
# Convert the integer into 
titanic[['male', 'Q', 'S']] = titanic[['male', 'Q', 'S']].astype(int)
```
