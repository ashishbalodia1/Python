```python
fruits = ["apple", "banana", "cherry"] 

# Print the first fruit.
print(fruits[0])

# Replace "banana" with "orange" 
fruits[1]="orange"
print(fruits)

# Print the length of the list.
print(len(fruits))
```

# list

```python
l=list(range(1,11))
print(l)
```

```python
# Using list comprehension
l=[ i for i in range( 1, 11 ) ]
print(l)
print(type(l))

print(l[0:3])
print(l[-3:])
```

```python
my_list =[5,2,9,1,7]

my_list.append(4)
print(my_list)

my_list.sort()
print(my_list)


my_list.remove(2)
print(my_list)

my_list.insert(1,99) 
print(my_list)


my_list.pop() 
print(my_list)

my_list.reverse() 
print(my_list)
```
