```python
# CAR GAME

print("Welcome to the Car Game")
# query=input(">").lower()     # All input command will be  in lower case 
query=""
isStarted= False
while True:
    query= input(">").lower()
    
    if query=='help':
        print("start - to star the car. \nstop- to stop the car. \naccelerate- to accelerate the car. \nquit- to leave the car .")
    
    elif query== 'start':
        if isStarted:
            print("Car is already started...")
        else:
            isStarted=True
            print("To start the car\n Car is starting...")
        
    elif query=="engine":
        print("Engine is starting...Started!")
    elif query=="accelerate":
        print("Accelerating the car...")

    
    elif query== 'stop':
        if not isStarted:
            print("Car is already stopped!")
        else:
            isStarted = False
            print("To stop the car\n Car stopped")
            
    elif query== "quit":
        break
    else:
        print("You entered a invalid query")
```

## Function

```python
# a block of code to perform repeative tasks
# Eg. input(), print(), max()
```

```python
def greet_user():
    print("Hii Everyone")
    print("Welcome abroad")


print("Start")
greet_user()
print("Finish")

```

```python
# Passing parameters to function
def greet_user(name):
    print(f"Hii {name}")
    print("Welcome to king's landing")


print("Start")
greet_user("John Snow")
greet_user("Mary Smith")

print("Finish")
```

```python
# Parameters vs arguments:

# Parameters are the variable names in a function's definition (placeholders).
# eg. name 

# Arguments are the actual values passed to the function when it is called. 
# Position and order matters
# eg.John Snow
```

```python
# Passing 2 parameters
def greet_user(first_name, last_name):
    print(f"Hii {first_name} {last_name}")
    print("Welcome to king's landing")


print("Start")
greet_user("John", "Snow")
greet_user("Mary", "Smith")

print("Finish")
```

```python
# Passing 2 parameters and changing the order of argument
def greet_user(first_name, last_name):
    print(f"Hii {first_name} {last_name}")
    print("Welcome to king's landing")


print("Start")
greet_user("Snow", "John")   # Order of argument is change-> will reflect in the output
greet_user("Mary", "Smith")

print("Finish")
```

```python
# keyword argumentation-> Position doesn't matters
def greet_user(first_name, last_name):
    print(f"Hii {first_name} {last_name}")
    print("Welcome to king's landing")


print("Start")
greet_user(last_name = "Snow", first_name="John")  
greet_user("Mary", "Smith")

print("Finish")

# Use keyword arguments to improve the readability of code specially while dealing with numericals
# Keyword arguments always comes after position arguments 
```

#### return statement

```python
# RETURN STATEMENT- send a value back to the caller but doesn't display anything
def square(number):
    return number * number


# method1
square(3)

# method2
# result=square(4)
# print(result)

#method3
print(square(5))  # Passing function call as an arguments for the print() function
```

```python
# if we don't use a return  statement in a function
def square(number):
    print(number * number)

print(square(3))
```

```python
# By default, python functions return None
# if it doesn't have a return statement at the end of the function body--> will return "None" means absence
```

#### Creating a reusable function

```python
# exe. Text to emoji converter using function
message= input(">")


def text_to_emoji(message):
    words= message.split(" ")
    emojis= {
        ":)": "😁",
        ":(": "😭"
    }
    output=""
    for word in words:
        output+= emojis.get(word, word) + " "
    return output


text_to_emoji(message)
# print(output)
```

```python
# A pure function should:
# 1. Take input as parameters
# 2. Return output using return
# 3. Not directly interact with user (no input() or print() inside)
```

```python
# A well-designed function should:
# 1. Do one specific task
# 2. Be independent
# 3. Avoid unnecessary side effects
# 4. Not mix computation with input/output unless required
```
