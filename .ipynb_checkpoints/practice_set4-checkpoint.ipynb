# Recursion


```python
# Program to write a recursive function that return factorial of a number
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n * fact(n-1)

try: 
    num=int(input("Enter the number: "))

    # Validate input
    if num<0:
        print("Error: Factorial is not defined for negative numbers!")

    # calculate factorial
    result= fact(num)
    print(f"The Factorial of {num} is {result}")

except Exception as e:
    print("Error: ",e)
```

```python
# Recursive function that return the sum of digits of a given numbers

def sumofdigits(n):
    if n==0:
        return 0
    
    return n % 10 + sumofdigits( n// 10)

try:
    num=int(input('Enter the number'))

    # function call
    result=sumofdigits(num)
    print(f"The sum of all the digits of a {num} is {result}")

except Exception as e:
    print("Error: ",e)
```

```python
# Program to calculate height of a pendulum when length and angle is given
import math

def heightofpendulum(length, angle_degree):
    angle_radians= math.radians(angle_degree)
    return length * (1- math.cos(angle_radians))

try:
    length=float(input("Enter the lenght of the pendulum: "))
    angle=float(input("Enter the angle of the pendulum: "))

    # calculation
    result=heightofpendulum(length, angle)
    print(f"The height of the pendulum is {result}")

except Exception as e:
    print("Error : ",e)
    
```

```python
def multiply(a, b):
    """
    Multiply two numbers and return the result.

    Parameters:
    a (int or float): The first number.
    b (int or float): The second number.

    Returns:
    int or float: The product of a and b.

    Raises:
    TypeError: If either a or b is not a number.
    """
    
    # Input validation
    # METHOD1
    # if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
    
    # METHOD2
    if not all(isinstance(x, (int, float)) for x in (a, b)):
        raise TypeError("Both arguments must be int or float.")
    
    return a * b

# METHOD 1
# Using help() to Displays docstring in a structured help format.
help(multiply)

print("---------------------------------------------------------------------------")

# METHOD 2
#  Using object.__doc__ to Directly prints the docstring text.
# print(multiply.__doc__)
```

```python
# FIBONACCI SERIES
# Using simple for loop

def fib(n):
    a,b=0,1
    for i in range(n):
        print(a, end=" ")
        a,b=b,a+b

fib(10)
```

```python
# Using while loop

def fib(n):
    a,b=0,1
    count=0
    while(count<n):
        print(a, end=" ")
        a,b= b, a+b
        count+=1
        
fib(8)
```

```python
# Using Recursive function (Not practical for large n)
def fibonacci(n):
    if n<=1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

for i in range(10):
    print(fibonacci(i), end=" ")
```

```python
# Using memoization(optimized Recursion)

from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

for i in range(11):
    print(fibonacci(i), end=" ")
```

```python
# Using List to store series
def finonacci(n):
    fib= [0,1]
    for i in range(2,n):
        fib.append(fibonacci[i-1] + fibonacci[i-2])   # Shamjh nhi aaya
    return fib[:n]

print(fibonacci(10))
```

```python
# Using Generator (Memory efficient)

def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

fib = fibonacci()

for _ in range(10):
    print(next(fib), end=" ")
```

```python
# Using map() and lambda function()

from functools import reduce

def fibonacci(n):
    return reduce(lambda x, _: x + [x[-1] + x[-2]], range(n-2), [0,1]) if n>1 else [0]

print(fibonacci(10))
    
```

```python
# 8️⃣ Using Matrix Exponentiation (Advanced / Fastest)
# Time Complexity: O(log n)
# Used in competitive programming.

def multiply(F, M):
    x = F[0][0]*M[0][0] + F[0][1]*M[1][0]
    y = F[0][0]*M[0][1] + F[0][1]*M[1][1]
    z = F[1][0]*M[0][0] + F[1][1]*M[1][0]
    w = F[1][0]*M[0][1] + F[1][1]*M[1][1]
    F[0][0], F[0][1], F[1][0], F[1][1] = x, y, z, w

def power(F, n):
    if n <= 1:
        return
    M = [[1, 1], [1, 0]]
    power(F, n//2)
    multiply(F, F)
    if n % 2 != 0:
        multiply(F, M)

def fibonacci(n):
    F = [[1, 1], [1, 0]]
    if n == 0:
        return 0
    power(F, n-1)
    return F[0][0]

for i in range(10):
    print(fibonacci(i), end=" ")
```

```python
# 9️⃣ Using Binet’s Formula (Mathematical)
# ⚠ May give floating-point inaccuracies for large n.
# Time complexity: O(1) and Space complexity: O(1)

import math

def fibonacci(n):
    phi = (1 + math.sqrt(5)) / 2
    return round((phi**n - (-phi)**(-n)) / math.sqrt(5))

for i in range(10):
    print(fibonacci(i), end=" ")
```
