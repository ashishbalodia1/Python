## Quick Guide to Numba

Namba is a just in time(JIT) compiler for pytohn that works best in the code that uses NumPy arrays and functions and loops.

Use through collection of decorators that can be applied to functions to instruct Numba to comiple them.

```python
from numba import jit
import logging
import numpy as np
import time

x = np.arange(100, dtype=np.float64).reshape(10, 10)

@jit
def go_fast(a): # Function is compiled to machine code when called the first time
    trace = 0
    for i in range(a.shape[0]):
        trace += np.tanh(a[i, i])
    return a + trace

# Do not report this... compilation time is included in the time of the execution
start = time.time()
go_fast(x)
end = time.time()
print("Elapsed (with compilation) = %s" % (end - start))


# Now the function is compiled, re-time it and executing from cache
start = time.time()
go_fast(x)
end = time.time()
print("Elapsed (after compilation) = %s" % (end - start))
```
