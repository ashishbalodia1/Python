```python
import logging
import random
```

```python
logging.basicConfig(filename= 'loggingerrors.txt', 
                    filemode='a',
                    format='%(asctime)s - %(levelname)s -',
                    datefmt='%Y-%m-%d %H:%M:%S')
```

```python
for i in range(0, 15):
    
    if i%2==0:
        logging.warning('Log warning message')
    elif i%3==0:
        logging.critical('Log critical message')
    else:
        logging.error('Log error message')
```
