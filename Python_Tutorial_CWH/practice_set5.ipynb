# Set

```python
my_set = {1, 2, 3, 3, 4}
print(my_set)               # Set will remove the duplicate items
```

```python
my_set = {1, 2, 3, 3, 4}
         
my_set.add(5)
print(my_set)
          
my_set.remove(2)
print(my_set)
          
my_set.discard(4)
print(my_set)
```

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
```

```python
# Write a program that takes a list of numbers and removes all duplicates using a set.

l1=[]
for i in range(1,10):
    l1.append(int(input("Enter a number")))
print("Original List: ",l1)

unique_list=list(set(l1))
print("List after removing duplicates", unique_list)
```

```python
products = {
    "Laptop": 55000.00,
    "Smartphone": 25000.00,
    "Headphones": 1500.00,
    "Keyboard": 1200.00,
    "Mouse": 800.00
}

# iterate over the dictionary
for product, price in products.items():
    print(f"{product}: {price}")

# Product with highest price
max_product = next(iter(products))
max_price = next(iter(products.values()))
for product, price in products.items():
    if max_price < price:
        max_price=price
        max_product=product
    
print(f'Product having maximum price {max_price} is {max_product}')
    
```

```python
# Merges two dictionaries into one
dict1 = {
    "Laptop": 55000.00,
    "Smartphone": 25000.00,
    "Headphones": 1500.00,
    "Keyboard": 1200.00,
    "Mouse": 800.00
}

dict2 = {
    "Aarav": "9876543210",
    "Meera": "9123456780",
    "Rohan": "9988776655"
}


merged_dict = dict1 | dict2

print("Dictionary 1:", dict1)
print("Dictionary 2:", dict2)
print("Merged Dictionary:", merged_dict)
```

```python
dict1 = {
    "Laptop": 55000.00,
    "Smartphone": 25000.00,
    "Headphones": 1500.00,
    "Keyboard": 1200.00,
    "Mouse": 800.00
}

dict2 = {
    "Aarav": "9876543210",
    "Meera": "9123456780",
    "Rohan": "9988776655"
}

merged_dict = dict1.copy()
merged_dict.update(dict2)

print(merged_dict)
```
