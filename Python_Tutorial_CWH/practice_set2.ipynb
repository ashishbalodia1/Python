# if else statement

```python
# Write a program that asks the user for a number and prints whether it is positive, negative, or zero.
try:
    num=int(input("Enter a number"))
    if num>0:
        print(f"{num} is a positive number")
    elif num<0:
        print(f"{num} is a negative number")
    elif num==0:
        print(f"{num} is neither positive nor a negative number. It's zero")
    
    else:
        print("invalid input")

except Exception as e:
    print("Error: ",e)
```

```python
# Create a program that checks if a person is eligible to vote (age >= 18).
try:
    age=int(input("Enter your age: "))
    if age>0 and age<18:
        print("Sorry You're just a child")
        
    elif age>=18 and age<100:
        print("Eligible for voting")
    else:
        print("Invalid range of age")

except Exception as e:
    print("Error: ",e)
```

```python
# Write a program that takes a number from the user and prints “Even” if it is
# even, otherwise “Odd”
try:
    num=int(input("Enter a number"))
    if num%2==0:
        print(f"{num} is a even number")
    else:
        print(f"{num} is a odd number")

except Exception as e:
    print("Error: ",e)
```

# Match case statements

```python
# Ask the user to enter a day number (1–7) and print the corresponding day of the week using 
# match case .

num=int(input("Enter a number for a specific day"))

match num:
    case 1: 
        print("Monday")
    case 2:
        print("Tueday")
    case 3:
        print("Wednesday")
    case 4: 
        print("Thursday")
    case 5: 
        print("Friday") 
    case 6:
        print("Saturday") 
    case 7:
        print("Sunday") 
        

```

```python
grade = input("Enter the grade")

match grade:
    case "A" | "A+":
        print("Excellent!")
    case "B" | "B+":
        print("Good job!")
    case "C":
        print("Average")
    case _:
        print("Needs improvement")
```

```python
# Write a program using match case that simulates a simple calculator.Ask the user for two
# numbers and an operation (+, -, *, /). Perform the operation using match case
n1=int(input("Enter the first number"))
n2=int(input("Enter the second number"))

op=input("Enter the operator to perform operation")

match op:
    case '+': 
        print(n1+n2)
    case '-': 
        print(n1-n2)
    case '*':
        print(n1*n2)
    case '/':
        print(n1/n2)
        
    case _:
        print("You have entered an Invalid operator!!")
        
```

```python
# Print the multiplication table of a number (entered by user).
num=int(input("Enter the number "))

print(f"Printing the multiplicatioin table of {num}") 
for i in range(1,11):
    print(f"{num} * {i} = {num*i}")
```

```python
# Use a while  loop to reverse a given number (e.g., 123 → 321)

num=123
rev=0

while(num>0):
    rem = num % 10
    rev = rev * 10 + rem
    num = num // 10
    
print(rev)

```

# Break, Continue and Pass statements

```python
# Use a for loop to print numbers from 1 to 10, but stop the loop if the number is 7 (use 
# break ).


for i in range(1,11):
    if i==7:
        break
    print(i)
```

```python
# Print numbers from 1 to 10, skipping the number 5 (use continue ).

for i in range(1,11):
    if i==5:
        continue
    print(i)
    
```

```python
# Write a loop that goes through numbers 1 to 5, but does nothing for number 3 
for i in range(1,11):
    if i==3:
        pass
    print(i)
```
