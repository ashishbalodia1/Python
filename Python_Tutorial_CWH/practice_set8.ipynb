# OS module

```python
import os

# Get the current working directory
current_dir = os.getcwd()
print("Current directory:", current_dir)

# List files and directories in a directory
files = os.listdir("..")# "." represents current directory
print("\nFiles in current directory:", files)

# Create a new directory
os.mkdir("../new_directory")
print("\nCreated new_directory")
```

# Creating Command Line Utilitie

```python
import argparse

parser = argparse.ArgumentParser(description="Count the number of lines in a file.")
parser.add_argument("filename", help="The file to count lines in.")

# ✅ Fix: use parse_known_args() for Jupyter compatibility
args, _ = parser.parse_known_args()

try:
    with open(args.filename, "r") as file:
        lines = file.readlines()
        print(f"Number of lines: {len(lines)}")
except FileNotFoundError:
    print(f"Error: File '{args.filename}' not found.")
```

```python
# search_word.py

import sys

def search_word(filename, target_word):
    try:
        with open(filename, "r") as file:
            content = file.read()
            return content.count(target_word)
    except FileNotFoundError:
        print("Error: File not found.")
        sys.exit(1)
    except Exception as e:
        print(f"Unexpected error: {e}")
        sys.exit(1)


def main():
    if len(sys.argv) != 3:
        print("Usage: python search_word.py <filename> <word>")
        sys.exit(1)

    filename = sys.argv[1]
    word = sys.argv[2]

    occurrences = search_word(filename, word)

    print(f"Word '{word}' appears {occurrences} times.")


if __name__ == "__main__":
    main()
```
