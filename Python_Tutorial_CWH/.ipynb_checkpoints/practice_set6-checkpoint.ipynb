# OOPs in Python

```python
class Car:
    def drive(self):
        return "Car is moving"

car1=Car()
print(car1.drive())
    
```

```python
class Person:
    def __init__(self, name, age):
        self.name=name
        self.age=age

    def info(self):
        return f"The name is {self.name} of age {self.age} years"

p1=Person("John", 22)
print(p1.info())
        
```

```python
class Animal:
    def sound(self):
        return f"Some sound"
        
class Dog(Animal):
    def sound(self):
        return super().sound() + " and Bark!"
        
dog1=Dog()
print(dog1.sound())
```

```python
# BANKING SYSTEM USING OOPs
class BankAccount:
    def __init__(self, account_holder, balance):
        if balance < 0:
            raise ValueError("Initial balance can't be negative")
            
        self.account_holder=account_holder
        self.balance=balance

    def deposit(self, amount):
    
        if amount <= 0:
            raise ValueError("Diposit amount must be positive")
        
        self.balance+= amount
        return self.balance
            
    def withdraw(self, amount):
        if amount<=0:
            raise ValueError("Withdraw amount must be positive")
            
        if amount > self.balance :
            raise ValueError("Insufficient balance")
        
        self.balance= self.balance - amount
        return self.balance

    def get_balance(self):
        return self.balance

        
class SavingsAccount(BankAccount):
    def __init__(self, account_holder, balance, interest_rate):
        if interest_rate < 0:
            raise ValueError("Interest rate can't be negative")
            
        super().__init__(account_holder, balance)   #PARENT CLASS constructor CALL
        self.interest_rate=interest_rate
      
    def apply_interest(self):
        interest_amount=round((self.balance * self.interest_rate)/100, 2)
        self.balance+=interest_amount
        return self.balance


class CurrentAccount(BankAccount):
    def __init__(self, account_holder, balance, overdraft_limit):
        if overdraft_limit < 0:
            raise ValueError("Overdraft  amount must be positive")
            
        super().__init__(account_holder, balance)
        self.overdraft_limit= overdraft_limit
        
    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")

        if self.balance - amount < -self.overdraft_limit:
            raise ValueError("Overdraft limit exceeded")

        self.balance = self.balance - amount
        return self.balance
        

# Testing
p1=SavingsAccount("JOhn", 1000 ,7)
print(p1.deposit(200))
print(p1.withdraw(200))
print( p1.get_balance())
print(p1.apply_interest())
        
p2=CurrentAccount("Maria", 100, 5000)
print(p2.withdraw(511))
print( p2.get_balance())
```

## Multiple inheritance

```python
class Vehicle:
    def __init__(self, brand, speed, fuel_type, **kwargs):
        self.brand = brand
        self.speed = speed
        self.fuel_type = fuel_type

    def drive(self):
        return "The vehicle is driving"

    def stop(self):
        return "The vehicle has stopped"


class ElectricVehicle(Vehicle):
    def __init__(self, battery_capacity, **kwargs):
        super().__init__(**kwargs)             
        self.battery_capacity = battery_capacity

    def charge(self):
        return "The battery is charging"

    def battery_status(self):                  
        return "The battery status is {}".format(self.battery_capacity)


class AutonomousVehicle(Vehicle):
    def __init__(self, autopilot_level, **kwargs):
        super().__init__(**kwargs)             
        self.autopilot_level = autopilot_level

    def enable_autopilot(self):
        return "Autopilot ON"

    def disable_autopilot(self):
        return "Autopilot OFF"


class TeslaCar(ElectricVehicle, AutonomousVehicle):
    def __init__(self, brand, speed, fuel_type, battery_capacity, autopilot_level, model):
        super().__init__(
            brand=brand,
            speed=speed,
            fuel_type=fuel_type,
            battery_capacity=battery_capacity,
            autopilot_level=autopilot_level
        )
        self.model = model

    def get_model(self):
        return self.model                    

    def __str__(self):
        return (f"Brand     : {self.brand}\n"
                f"Speed     : {self.speed}\n"
                f"Fuel Type : {self.fuel_type}\n"
                f"Battery   : {self.battery_capacity}\n"
                f"Autopilot : {self.autopilot_level}\n"
                f"Model     : {self.model}")


# Testing
car1 = TeslaCar("Tesla", "250km/h", "Electric", "100 kWh", "Level-4", "Model-S")
print(car1.get_model())
print(car1)
print(car1.drive())
print(car1.charge())
print(car1.battery_status())
print(car1.enable_autopilot())
print(TeslaCar.__mro__)
```
