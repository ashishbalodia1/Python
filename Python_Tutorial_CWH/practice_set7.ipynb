```python
def timer(fx):
    def wrapper():
        print("Function call starting")
        fx()
        print("Function call ended")  
    return wrapper

@timer
def sumofnumbers():
    count=0
    for i in range(1,100):
        count+=i
    return count

print(sumofnumbers())
```

```python
import time

def timer(fx):
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        
        result = fx(*args, **kwargs)
        
        end_time = time.perf_counter()
        execution_time = end_time - start_time
        
        print(f"Execution time: {execution_time:.6f} seconds")
        
        return result
    return wrapper


@timer
def sumofnumbers():
    count=0
    for i in range(1,100):
        count+=i
    return count

print(sumofnumbers())
```

## Getter and Setter


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self._age = age  # Convention: _age indicates it's intended to be "private"
    
    def get_age(self): # Getter for age
        return self._age

    def set_age(self, new_age): # Setter for age
        if new_age >= 0 and new_age <= 150: # Validation
            self._age = new_age
        else:
            print("Invalid age!")

# Testing
person = Person("Alice",30)
print(person.get_age()) # Output: 30

person.set_age(35)
print(person.get_age()) # Output: 35

person.set_age(-5) # Output: Invalid age!
print(person.get_age()) # Output: 35 (age wasn't changed)
```

```python
# USING @PROPERTY DECORATOR for getter

class Person:
    def __init__(self, name, age):
        self.name = name
        self._age = age       # Convention: _age for "private" attributes
    
    @property        # This makes 'age' a property (the getter)
    def age(self):
        return self._age
    
    @age.setter           # This defines the setter for the 'age' property
    def age(self, new_age):
        if new_age >= 0 and new_age <= 150:
            self._age = new_age
        else:
            print("Invalid age!")
    
person = Person("Bob",40)
print(person.age) # Output: 40  (Looks like direct attribute access, but calls the getter)
person.age =45 # (Calls the setter – looks like attribute assignment)
print(person.age)
# del person.age    # AttributeError: property 'age' of 'Person' object has no deleter
person.age =-22 #Output: Invalid age
```

```python
# Read-Only Properties 
class Circle:
    def __init__(self, radius):
        self._radius = radius
        
    @property
    def radius(self):
        return self._radius
        
    @property
    def area(self): # Read-only computed property
        return 3.1416 * self._radius * self._radius

c = Circle(5)
print(c.radius) # 5
print(c.area) # 78.54
# c.radius = 10  # Raises AttributeError: can't set attribute
# c.area = 20 # Raises AttributeError: can't set attribut
```

```python
class Employee:
    def __init__(self, salary):
        if salary <0:
            raise ValueError("Salary can't be negative")
        self._salary=salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, new_salary):
        if new_salary < 0:
            raise ValueError("New salary can't be negative")
        self._salary = new_salary

e1=Employee(10_000_000)
print(e1.salary)
e1._salary=-220000   # Will raise a Value Error
```

# *args and **kwarg

```python
# *args :- Allows you to pass a variable number of positional arguments into a tuple.

def my_function(*args):
    print(type(args)) # <class 'tuple'>
    for arg in args:
        print(arg)
        
my_function(1, 2, 3, "hello") # Output: 1 2 3 hello

my_function() # No output (empty tuple)

my_function("a","b") # Output: a b
```

```python
# **kwargs :- : Allows you to pass a variable number of keyword arguments into a dictionary

def my_function(**kwargs):
    print(type(kwargs)) # <class 'dict'>
    for key, value in kwargs.items():
        print(f"{key}: {value}")

my_function(name="Alice", age=30, city="New York")
# Output:
# name: Alice
# age: 30
# city: New York

my_function() # No output (empty dictionary)

my_function(a=1, b=2)
# Output:
# a: 1
# b: 
```

```python
import time
import functools

def monitor(func):
    @functools.wraps(func)   # Preserves metadata
    def wrapper(*args, **kwargs):
        print(f"\n[MONITOR] Running: {func.__name__}")
        print(f"[MONITOR] Positional args: {args}")
        print(f"[MONITOR] Keyword args: {kwargs}")

        start_time = time.time()

        try:
            result = func(*args, **kwargs)
            end_time = time.time()

            print(f"[MONITOR] Result: {result}")
            print(f"[MONITOR] Execution Time: {end_time - start_time:.6f} seconds")

            return result

        except Exception as e:
            print(f"[MONITOR] Exception occurred: {e}")
            raise   # Re-raise so error is not swallowed

    return wrapper


@monitor
def calculate_weighted_score(student_name, *scores, weight=1.0, bonus=0):
    if not scores:
        raise ValueError("At least one score must be provided")

    average = sum(scores) / len(scores)
    final_score = (average * weight) + bonus

    return {
        "student": student_name,
        "final_score": round(final_score, 2)
    }


# TESTING 
# Normal case
calculate_weighted_score("Rahul", 80, 90, 99, 50, weight=1.1, bonus=5)

```

```python
# Adding and subjtracting two 3D vectors
class Vector:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

        
    #-------------------------------------
    # Addition
    #-------------------------------------
    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(
                self.x + other.x,
                self.y + other.y,
                self.z + other.z)
            
        elif isinstance( other, (int, float)):
            return Vector(
                self.x + other,
                self.y + other,
                self.z + other)
            
        return NotImplemented

    # -----------------------------
    # Subtraction 
    # -----------------------------

    def __sub__(self, other ):
        if isinstance(other, Vector):
            return Vector(
                self.x - other.x,
                self.y - other.y,
                self.z - other.z)
        elif isinstance(other, (int, float)):
            return Vector(
                self.x - other,
                self.y - other,
                self.z - other)

        return NotImplemented 

    
    def __radd__(self, other):
        if other == 0:
            return self
            
        # For symmetry, delegate to __add__
        return self.__add__(other)

    def __repr__(self):
        return f"Vector({self.x}, {self.y}, {self.z})"

v1 = Vector(1,2,3)
v2 = Vector(4,5,6)

print("Addition: ", sum([v1, v2]))
print("Substtraction: ", v1-v2)
```

```python
class Vector:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

        
    #-------------------------------------
    # Addition
    #-------------------------------------
    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(
                self.x + other.x,
                self.y + other.y,
                self.z + other.z)
            
        elif isinstance( other, (int, float)):
            return Vector(
                self.x + other,
                self.y + other,
                self.z + other)
            
        return NotImplemented

    def __radd__(self, other):
        if other == 0:
            return self
            
        # For symmetry, delegate to __add__
        return self.__add__(other)

    # -----------------------------
    # Subtraction 
    # -----------------------------

    def __sub__(self, other ):
        if isinstance(other, Vector):
            return Vector(
                self.x - other.x,
                self.y - other.y,
                self.z - other.z)
        elif isinstance(other, (int, float)):
            return Vector(
                self.x - other,
                self.y - other,
                self.z - other)

        return NotImplemented 

    
    def __rsub__(self, other):
        if other == 0:
            return self
            
        # For symmetry, delegate to __add__
        return self.__sub__(other)

    #--------------------------------
    # Multiplication
    #--------------------------------
    def __mul__(self, other ):
        if isinstance(other, Vector):
            return Vector(
                self.x * other.x,
                self.y * other.y,
                self.z * other.z)
        elif isinstance(other, (int, float)):
            return Vector(
                self.x * other,
                self.y * other,
                self.z * other)

        return NotImplemented 

    
    def __rmul__(self, other):
        if other == 0:
            return self
            
        # For symmetry, delegate to __add__
        return self.__mul__(other)

    # ----------------------
    # Divsion
    #----------------------
    def __truediv__(self, other ):
        if isinstance(other, Vector):
            if other.x == 0 or other.y == 0 or other.z == 0:
                    raise ValueError("Cannot divide by a Vector with a zero component")
            return Vector(
                self.x / other.x,
                self.y / other.y,
                self.z / other.z)
        elif isinstance(other, (int, float)):
            return Vector(
                self.x / other,
                self.y / other,
                self.z / other)

        return NotImplemented 
    
    def __rtruediv__(self, other):
        if other == 0:
            return self
            
        # For symmetry, delegate to __add__
        return self.__truediv__(other)

    
    # ----------------------------
    # String Representation
    #-----------------------------

    def __repr__(self):
        return f"Vector({self.x}, {self.y}, {self.z})"

    def __str__(self):
        return f"({self.x}, {self.y}, {self.z})"


v1 = Vector( 10, 55, 30)
v2 = Vector( 4, 5, 5)

print("Addition: ", sum([v1, v2]))
print("Substtraction: ", v1-v2)
print("Multiplication: ", v1 * v2)
print("Divsion:  ", v1 / v2)
```

```python
import logging

# PART 1: Logging Configuration
logging.basicConfig(
    level=logging.ERROR,
    format="%(asctime)s | %(levelname)s | %(message)s",
    handlers=[
        logging.FileHandler("errors.log"),
        logging.StreamHandler()
    ]
)

# PART 2: Custom Exceptions
class InvalidInputError(Exception):
    pass

class NegativeNumberError(Exception):
    pass

class EmptyInputError(Exception):
    pass


# PART 3: Validation Function
def valid_input(value):

    if value == "":
        raise EmptyInputError("Input can't be empty")

    try:
        number = float(value)
    except ValueError:
        raise InvalidInputError(f"'{value}' is not a valid number!")

    if number < 0:
        raise NegativeNumberError(f"{number} can't be negative")

    return number


# PART 4: Main Program
def main():
    while True:
        raw = input("Enter a positive number (or 'quit'): ").strip()

        if raw.lower() == "quit":
            break

        try:
            number = valid_input(raw)
            square = number ** 2

            print(f"Number : {number}")
            print(f"Square : {square}")

        except EmptyInputError as e:
            logging.error(f"EmptyInputError     → {e}")
            print("Please type something!")

        except InvalidInputError as e:
            logging.error(f"InvalidInputError   → {e}")
            print("Only numbers allowed!")

        except NegativeNumberError as e:
            logging.error(f"NegativeNumberError → {e}")
            print("Positive numbers only!")

    print("\nBye! Check errors.log for history.\n")


if __name__ == "__main__":
    main()
```
