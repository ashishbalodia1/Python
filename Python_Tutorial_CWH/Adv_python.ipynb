# List comprehension

```python
# Provides a concise way to create lists
```

```python
lst1=[]
def lst_square(lst):
    for i in lst:
        lst1.append(i*i)
    return lst1
```

```python
lst_square([1,2,3,4,5,6,7,8])
```

```python
lst=[1,2,3,4,5,6,7,8]
square = [i*i for i in lst]
```

```python
square
```

## Iterables vs Iterators

```python
# An iterable is any object you can loop over — like a list, tuple, string, etc.
# List are iterable 
lst= [1,2,3,4,5,6]

for i in lst:
    print(i)

# Iterating through the list means iterating through memeory locations---+
# Python internally iterates through memory locations of the list elements one by one
```

```python
# An iterator is an object that:
# Holds a stream of values
# Remembers where it is in the sequence
# Gives you the next value only when asked (via) next()
```

```python
# iter() function converts a list( iterable) into iterator
iter(lst)
# All the values are not initialize in the memory all at once unless we call an in built function next()
# Only when we call next() does the first element get initialized in memory
```

```python
# Method1- Manual interation using next()
```

```python
# So converting list which is  iterable into iterator
lst1= iter(lst)
# lst1 is now iterator

lst1
```

```python
# Reterieve the elements inside iterator
next(lst1)    # Return 1
next(lst1)    # Return 2 

# ONly one element is initialized at the memory at the time of calling the next() function. 
# Once all elements are exhausted, calling next() again raises a StopIteration exception —
# Python's way of saying "nothing left to give!"
```

```python
# Method2- Using a for loop
```

```python
for i in lst1:
    print(i)


# StopIteration Exception is already handled in this for loop by default.
```

### Asynchronous Execution

```python
# Launching parallel execution
```

***ThreadPoolExecuter***

```python
import time
from concurrent.futures import ThreadPoolExecutor

def returnnumber(a):
    time.sleep(1)
    return a 
```

```python
returnnumber(2)
```

```python
start = time.time()
with ThreadPoolExecutor(max_workers = 5) as executor:
    for result in executor.map(returnnumber, range(100)):
        print("Count: {0}:".format(result))

print("The total time is : {0}".format(time.time()-start))
```

## Zip funciton- Parallel iteration

1. Allows iteration through multiple  iterables.
2. It's an iterable So can use for loop or next() to iterate.

```python
# Used in Data preprocessing and feature engineering.
```

```python
l1= ['krish', 'sam', 'john', "Arry"]
l2=['a', 'b', 'c']
```

```python
for i in l1:
    print(i)
```

```python
output = zip(l1, l2)
```

```python
output
```

```python
list(output)
```

```python
#### Zip object is an iterator
next(output)


# StopIteration error dega
```

```python
l1= ['krish', 'sam', 'john', "Arry"]
l2=['a', 'b', 'c']
l3= [1,2,3]
```

```python
output = zip(l1, l2, l3)
for i, j, k in output:
    print(i, j, k)
```

```python
# Create empty zip object
list(zip())
```
