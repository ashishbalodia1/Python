```python
from collections import namedtuple

from scipy.sparse.csgraph import breadth_first_tree


class Dog:
    def __init__(self, *args):
       if len(args) == 1:
           self.name=args[0]
       elif len(args) == 2:
           self.name=args[0]
           self.bread=args[1]
       elif len(args) == 3:
           self.name=args[0]
           self.bread=args[1]
           self.age= args[2]
```

```python
def make_sound(self, sound):
    return "The animal is {} and says {}".format(self.name,sound)
```

```python
dog = Dog("bruno", "mammale", 4)
```
