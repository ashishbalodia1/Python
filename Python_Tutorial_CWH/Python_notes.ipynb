```python
# exe. Text to emoji converter using function
message= input(">")


def text_to_emoji(message):
    words= message.split(" ")
    emojis= {
        ":)": "😁",
        ":(": "😭"
    }
    output=""
    for word in words:
        output+= emojis.get(word, word) + " "
    return output


text_to_emoji(message)
# print(output)
```

```python
# Functions that operate on other functions are called "higher-order functions." 
```

```python
def mult_by_five(x):
    return 5 * x

def call(fn, arg):
    """Call fn on arg"""
    return fn(arg)

def squared_call(fn, arg):
    """Call fn on the result of calling fn on arg"""
    return fn(fn(arg))

print(
    call(mult_by_five, 1),
    squared_call(mult_by_five, 1), 
    sep='\n', # '\n' is the newline character - it starts a new line
)
```

```python
# Using the max function 
def mod_5(x):
    """Return the remainder of x after dividing by 5"""
    return x % 5

print(
    'Which number is biggest?',
    max(100, 51, 14),
    'Which number is the biggest modulo 5?',
    max(100, 51, 14, key=mod_5),
    sep='\n',
)
```

```python
class Animal:            # Parent class
    location="Australia"        # class instance/ variable

    def __init__(self, name):       # constructor
        self.name=name              # instance variable

    def speak(self):               # Parent class method
        print("Generic animal sound")

class Dog(Animal):               # chlid class 1
    def speak(self):
        super().speak()         # Acceessing method of parent class
        print("Woaf")

class Cat(Animal):           # chlid class 1
    def speak(self):
        super().speak()       # Acceessing method of parent class
        print("Meow")

d1=Dog("Bruno")    # Creating instance of Dog class
d1.speak()

c1=Cat("Meemee")  # Creating instance of Cat class
c1.speak()
```

## Single inheritance

```python
class Car:
    color="Black"

    @staticmethod
    def start():
        print("Car started...")

    @staticmethod
    def stop():
        print("Car stopped")

class ToyataCar(Car):
    def __init__(self, name):
        self.name=name

car1=ToyataCar("Fortuner")
car2=ToyataCar("Prius")

print(car1.color)
```

## Multi level inheritance 

```python
class Car:
    def __init__(self, type):
        self.type="Electric"
        
    # color="Black"

    @staticmethod
    def start():
        print("Car started...")

    @staticmethod
    def stop():
        print("Car stopped")

class ToyataCar(Car):
    def __init__(self, name, brand):
        super().__init__(type)
        self.name=name
        self.brand=brand

class Fortuner(ToyataCar):
    def __init__(self, type):
        self.type=type
        
car1=Fortuner("Diesel")
car1.start()

car2=ToyataCar("Prius", "Electric")
print(car2.type)
```

## Multiple inheritance

```python
class Parent1:
    def method1(self):
        print("Method from parent 1")
        
class Parent2:
    def method2(self):
        print("Method from parent 2")
        
class Child(Parent1, Parent2):
    def method3(self):
        print("Method from child")

obj=Child()
obj.method1()
obj.method2()
obj.method3()
```

# Access Modifiers

```python
# PUBLIC ACCESS

class Employee:
    def __init__(self):
        self.name="Harry"

a=Employee()
print(a.name)
```

```python
# PRIVATE ACCESS
#  double underscore for private
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary   # Private variable

    def show_salary(self):
        print("Salary :", self.__salary)

e = Employee("Rohan", 50000)

print(e.name)          # This will work as its a public variable

# print(e.__salary)      # This will give AttributeError

e.show_salary()        # Way to access private variable
```

```python
# PROTECTED ACCESS MODIFER
# single underscore for protected
class Student:
    def __init__(self, name):
        self._name = name   # Protected variable

    def funName(self):
        print(f"The name is {self._name}")


class Subject(Student):
    pass


# Creating objects
obj1 = Student("Harry")
obj2 = Subject("Rohan")

# Accessing protected member
print(obj1._name)        # Accessible
obj1.funName()

print(obj2._name)        # Accessible in subclass
obj2.funName()
```

## List comprehension

```python
# SQUARE OF NUMBER
numbers=[1,2,3,4,5,6]

'''
sq=[]
for number in numbers:
    sq.append(number*number)
print(sq)
'''

# LIST COMPREHENSION
sq=[number*number for number in numbers]
print(sq)
```

```python
# EVEN NUMBER
numbers=[1,2,3,4,5,6]
'''
even_num=[]
for num in numbers:
    if num%2==0:
        even_num.append(num)
print(even_num)
'''

# LIST COMPREHENSION
even_num=[num for num in numbers if num%2==0]
print(even_num)
```

```python
# Only include results greater than 50
numbers=[53,25,98,1,56,2,670,7,23,6]
'''
g_num=[]
for number in numbers:
    if number>50:
        g_num.append(number)
print(g_num)'''

g_num=[number for number in numbers if number>50]
print(g_num)
```
