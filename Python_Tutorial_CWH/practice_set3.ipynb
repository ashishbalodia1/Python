# String Formatting

```python
# Using format()
name="john"
age=25

print("My name is John and I am 25 years old.".format(name, age))
```

```python
# Using f-string
name="John"
age=33
print(f"My name is {name} and I am {age} years old.")
```

#  String Manipulation Challenges

```python
sentence = "Coding in Python is fun" 
sentence=sentence.replace("fun", "awesome")
print(sentence)
```

```python
# Index of the word 'Python' in the sentence
sentence = "Coding in Python is fun"
sentence.find("Python")
```

```python
# Convert to upper case and print it 
sentence = "Coding in Python is fun"
sentence = sentence.upper()
print(sentence)
```

# Bonus Questions

```python
# Write a program that counts how many vowels are in a given string.
str = "Coding in Python is fun"
vowel_count=0
for i in str:
    if i.lower() in "aeiou" :
        vowel_count+=1
        
print(vowel_count)
```

```python
# Take a user input string and check if it is a palindrome (same forwards and backwards)
str = "pop"
if str==str[::-1]:
    print("It's a palindrome")
else:
    print("Not a palindrome")
```
